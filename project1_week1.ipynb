```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
import statsmodels.formula.api as smf

hb = pd.read_csv("dataset/handball_wide.csv")
hb = hb.rename(columns={"bet.probs.diff": "betdiff"})
```

```python
m1 = smf.ols("scorediff ~ betdiff + season", data=hb).fit()
print(m1.summary())
```

```python
fitted    = m1.fittedvalues
resid     = m1.resid
std_resid = m1.get_influence().resid_studentized_internal

fig, ax = plt.subplots(2, 2, figsize=(11, 9))

sns.scatterplot(x=fitted, y=resid, ax=ax[0, 0], alpha=0.4)
ax[0, 0].axhline(0, color="red")
ax[0, 0].set_title("Residuals vs Fitted")

sm.qqplot(std_resid, line="45", ax=ax[0, 1])
ax[0, 1].set_title("Normal Q-Q")

sns.scatterplot(x=fitted, y=np.sqrt(np.abs(std_resid)), ax=ax[1, 0], alpha=0.4)
ax[1, 0].set_title("Scale-Location")

sm.graphics.influence_plot(m1, ax=ax[1, 1], criterion="cooks", alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
hb["pred1"]    = m1.predict(hb)
hb["abs_err1"] = (hb["scorediff"] - hb["pred1"]).abs()

print("Model 1 | mean abs diff:  ", hb["abs_err1"].mean())
print("Model 1 | median abs diff:", hb["abs_err1"].median())
```

```python
sns.boxplot(y=hb["abs_err1"])
plt.title("Absolute difference between true and predicted scorediff (model 1)")
plt.show()
```

```python
m2 = smf.ols("scorediff ~ betdiff + C(hometeam) + C(awayteam)", data=hb).fit()
hb["pred2"]    = m2.predict(hb)
hb["abs_err2"] = (hb["scorediff"] - hb["pred2"]).abs()

print("Model 1 | mean:", hb["abs_err1"].mean(), "| median:", hb["abs_err1"].median())
print("Model 2 | mean:", hb["abs_err2"].mean(), "| median:", hb["abs_err2"].median())

sns.boxplot(data=hb[["abs_err1", "abs_err2"]])
plt.ylabel("Absolute difference")
plt.show()
```
